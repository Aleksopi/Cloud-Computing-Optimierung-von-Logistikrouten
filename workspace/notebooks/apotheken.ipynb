# Projekt 9: Optimierung von Logistikrouten

### Problemdefinition

Wir sind ein grösserer Logistikanbieter. Wir beliefern 400 Schweizer Apotheken (`/workspace/data/apotheken_400.geojson`)
Wir haben einen Hauptsitz in Bern, sowie 4 Verteilzentren in der Nähe von Basel, Zürich, Genf und Chur, sowie weitere kleinere Verteilzentren (mVZ).

Unsere Verteilzentren und unser Hauptquatier sind mit einer effizienten Zuganbindung ausgestattet, 60% der Quellwaren kommen direkt vom Hauptquatier und müssen an das Zielverteilzentrum versandt werden, 40% der Quellwaren sind in einem Verteilzentrum. Davon wiederum sind 70% für eine Weiterlieferung an ein anderes Verteilzentrum oder an ein Hauptquatier bestimmt, 30% wiederum gehen direkt an ein kleineres Verteilzentrum oder direkt an eine Apotheke.


### Aufgaben der Logistik

Unsere Logistikprogramme müssen
- In Schritt 1: Darstellen, wie Waren verteilt werden, sodass diese alle im Ziel-VZ (Verteilzentrum) ankommen und die optimalen Standorte für die mVZ bestimmen.
- In Schritt 2: Darstellen, welche Apotheke im Einflussbereich eines gegebenen VZ oder mVZ sind, eine Apotheke wird nur von einem VZ oder mVZ beliefert.
- In Schritt 3: Den Bedarf der jeweiligen Apotheken festlegen. 1 Ware pro 8 000 - 15 000 Einwohner (variabler Konsum).
- In Schritt 4: Darstellen, welche Lieferwägen welche Routen fahren müssen. Jedes VZ hat ein bis drei LKW mit 600km Maximalreichweite, sowie 5 bis 12 EVans mit 150km Maximalreichweite. Dabei darf ein Lieferwagen maximal 8h unterwegs sein und muss dann innerhalb der 8h wieder zurück am Start sein. Das beliefern einer Apotheke dauert 40 Minuten mit einem LKW und 30 Minuten mit einem EVan.
    - Sub-Schritt 1: Waren von HQ und VZ zu allen VZ und mVZ verteilen (jeweils die kürzeste Route pro ware (mehrmals Djikstra))
    - Sub-Schritt 2: Die Waren vom jeweiligen VZ/mVZ an die Apotheken verteilen
- Ausserdem: Unsere Kosten berechnen, um die Preise korrekt zu Berechnen:
    - HQ zu VZ: 10 CHF pro Ware und 100km Luftlinie
    - VZ zu VZ: 12 CHF pro Ware und 100km Luftlinie
    - VZ zu mVZ: 15 CHF pro Ware und 100km Luftlinie
    - VZ/mVZ zu Apotheke: 1 Stunde kostet 50 CHF und 100km kosten 30 CHF für den LKW und 20 CHF für den EVan

```python
# Pfad zu Ihren GeoJSON-Daten festlegen
geojson_path = Path("../../workspace/data/apotheken_400.geojson")

print(f"Lade Apotheken-Daten von {geojson_path}...")
with open(geojson_path, "r", encoding="utf-8") as f:
    geojson_data = json.load(f)

# Features in ein flaches Pandas DataFrame konvertieren
records = []
for feature in geojson_data.get("features", []):
    props = feature.get("properties", {}).copy()
    geom = feature.get("geometry", {})
    if geom and geom.get("type") == "Point":
        lon, lat = geom.get("coordinates")
        props["lat"] = lat
        props["lon"] = lon
        records.append(props)

df_pharmacies = pd.DataFrame(records)
print(f"✅ {len(df_pharmacies)} Apotheken erfolgreich geladen!")
```

### Schritt 1

#### Code -- alle VZ / mVZ

```python
# 1. Map-Instanz für die Session erstellen
map = map_object.new(center_lat=46.95, center_lon=7.45, zoom=8)

# 2. Locations planen und in die Karte integrieren
# (Hier werden die Apotheken-Daten 'df_pharmacies' aus deinem vorherigen Schritt genutzt)
vz_locs, mini_locs, assignments, hub_to_pharm = plan_locations(
    df_pharmacies=df_pharmacies,  # Stelle sicher, dass dies dein geladenes DataFrame ist
    hq_lat=46.95, 
    hq_lon=7.45, 
    map_obj=map.drawInterface # Pass the interface
)

# 3. Alle Apotheken als Punkte auf der Karte hinzufügen (optional, zur besseren Übersicht)
for idx, row in df_pharmacies.iterrows():
    map.drawInterface.add_point(
        lat=row['lat'], 
        lon=row['lon'], 
        layer="Apotheken", 
        color="#555555", 
        tooltip=f"ID: {idx}"
    )

map_locations_saved = map.save()
```

#### Karte aller VZ / mVZ

```python
# Karte anzeigen
map.render()
```

#### Code -- Direkteinfluss

```python
map_infl_old = map_object.load(map_locations_saved)

def _influence(assignments, hub_locations, pharmacy_df, map_obj):
    logger.info("[Step 1] Zeichne Einflusslinien auf die Karte...")
    
    for pid, hub_name in assignments.items():
        # Apotheken-Koordinaten (wir nutzen die korrigierten Spaltennamen)
        p_lat = pharmacy_df.loc[pid, 'lat']
        p_lon = pharmacy_df.loc[pid, 'lon']
        
        # Hub-Koordinaten aus dem dict der Standorte
        if hub_name in hub_locations:
            h_lat, h_lon = hub_locations[hub_name]
            
            # Farbe je nach Hub-Typ (VZ = Blau, mVZ = Grün)
            color = "blue" if "VZ_" in hub_name and "mVZ" not in hub_name else "green"
            
            # Linie zeichnen
        map_obj.add_line(
            start=(h_lat, h_lon), 
            end=(p_lat, p_lon), 
            color=color,
            weight=2,
            opacity=0.9
        )
    
    logger.info("[Step 1] Alle Linien gezeichnet.")

all_hub_locs = {**vz_locs, **mini_locs}

# Berechne und visualisiere
_influence(
    assignments=assignments, 
    hub_locations=all_hub_locs, 
    pharmacy_df=df_pharmacies, 
    map_obj=map_infl_old.drawInterface
)
```

### Schritt 2

Berechne Einflussbereiche der Verteilzentren, um deren Warenfluss festzulegen (typische Logistikplanung)

#### Code

```python
# Karte aus Schritt 1 laden — enthält bereits alle VZ/mVZ/HQ und Apotheken-Marker
map_influence = map_object.load(map_locations_saved)

# Nur VZ + mVZ als Einfluss-Hubs übergeben
# (HQ beliefert keine Apotheken direkt: "eine Apotheke wird nur von einem VZ oder mVZ beliefert")
hub_locs = {**vz_locs, **mini_locs}

# Apotheken zuweisen + Fahrtrouten auf Karte zeichnen (Optimiert auf kürzeste Zeit)
final_assignments = calculate_influence(
    hub_locations = hub_locs,
    pharmacy_df   = df_pharmacies,
    map_obj       = map_influence.drawInterface,
    metric        = "time",         # Enforces always shortest time metric
    max_workers   = 1,              # Enforces stable sequential OSRM calls
)

# Karten-Zustand für spätere Schritte speichern
map_influence_saved = map_influence.save()

print(f"✅ {len(final_assignments)} Apotheken zugewiesen")
```

### Schritt 4

```python
# ─────────────────────────────────────────────────────────────────────────────
# Schritt 4 — Routenplanung
# ─────────────────────────────────────────────────────────────────────────────

# Karte aus Schritt 2 als Basis laden
map_routes = map_object.load(map_influence_saved)

# Hubs-DataFrame aus Schritt-1-Ergebnissen zusammenstellen
hubs_records = [{"hub_name": "HQ", "lat": 46.95, "lon": 7.45, "type": "HQ"}]

for name, (lat, lon) in vz_locs.items():
    hubs_records.append({"hub_name": name, "lat": lat, "lon": lon, "type": "VZ"})

for name, (lat, lon) in mini_locs.items():
    hubs_records.append({"hub_name": name, "lat": lat, "lon": lon, "type": "mVZ"})

df_hubs = pd.DataFrame(hubs_records)

# assigned_hub aus Schritt 2 in pharmacy DataFrame eintragen
if "assigned_hub" not in df_pharmacies_ranked.columns:
    df_pharmacies_ranked = df_pharmacies_ranked.copy()
    df_pharmacies_ranked["assigned_hub"] = (
        df_pharmacies_ranked.index.map(final_assignments)
    )

# Einziger Aufruf ins Step-4-Modul
routing_results = plan_routes(
    pharmacy_df = df_pharmacies_ranked,
    hubs_df     = df_hubs,
    map_obj     = map_routes.drawInterface,
    max_workers = 16,
)

# Karte für spätere Schritte speichern
map_routes_saved = map_routes.save()

# Kennzahlen für Weiterverarbeitung
metrics     = routing_results["last_mile_delivery"]["metrics"]
hub_metrics = routing_results["hub_distribution_network"]
```
